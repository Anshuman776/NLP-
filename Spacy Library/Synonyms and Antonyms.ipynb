```python
from nltk.corpus import wordnet
```

```python
syn = wordnet.synsets('Book')
```

```python
print(syn[0].definition())
```

**Synonyms**

```python
synonyms =[]
```

```python
for s in wordnet.synsets('Computer'):
    for l in s.lemmas():
        synonyms.append(l.name())
```

```python
print(synonyms)
```

so we got the synonyms for computer

```python
synonyms =[]
for s in wordnet.synsets('strong'):
  for l in s.lemmas():
    synonyms.append(l.name())
```

```python
print(synonyms)
```

**Antonyms**

```python
ant =[]
for a in wordnet.synsets('Healthy'):
  for l in a.lemmas():
    if l.antonyms():
      ant.append(l.antonyms()[0].name())
```

```python
print(ant)
```
