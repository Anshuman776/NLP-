```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
s = nlp('AKGEC is looking for data Science student')
for token in s:
  print(token.text)
```

```python
s =nlp(u'The cost of Iphone in U.K is 699 \ $')
for token in s:
  print(token.text)
```

Even though the code runs, Python is warning you that \$ is meaningless.

```python
# pos_ -it indicates part of speech
for token in s:
  print(token.text,token.pos_)
```

## it basically tell the part of speech like noun , pronoun , number

Getting the dependency parsing with dep_ function

```python
s = nlp(u"He isn't going to     play today")
for token in s:
  print(token.text,token.pos_,token.dep_)
```

is and n't are being categorised differently 'n't' is being treated as negation
Due to the extra space between to and play we can see in the output a space is also being categorised

**sentence Tokenization**

```python
s = nlp(u'This is the first sentence. I gave give fullstop please check.Let us study now' )
for sentence in s.sents:
  print(sentence)
```
