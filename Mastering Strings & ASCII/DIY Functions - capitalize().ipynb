**The capitalize Function**
it is bulid in function , it changes the string into proper formate 

```python
text = "the quick brown fox jumps over the LAZY dog"
print(text.capitalize())
```

creating the capitalize function by step by step

```python
first_char = text[0].upper() # Capitalizing the first letter 
rest_of_text = text[1:].lower() # Lowercasing the rest of the string 
result = first_char + rest_of_text
print(result)
```

We use upper() to capitalize the first character.

We use lower() to convert all other characters to lowercase.
