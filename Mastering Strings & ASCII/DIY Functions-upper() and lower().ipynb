```python
# Now if we want to do the same thing manually without using the upper and lower function 
```

```python
def upper(text):
    string=''
    for char in text :
        if ord(char)>=97 and ord(char)<=122:
            #convert to upper by subtracting 32 from ASCII value
            string +=chr(ord(char)-32)
        else:
            # add the character as is if it's not a lowercase letter 
            string+=char
    return string

        
```

```python
print(upper('delhi'))
```

```python
def lower(text):
    string=''
    for char in text :
        if ord(char)>=65 and ord(char)<=90:
            #convert to lower by adding 32 from ASCII value
            string +=chr(ord(char)+32)
        else:
            # add the character as is if it's not a uppercase letter 
            string+=char
    return string
```

```python
print(lower('ANSHUMAN'))
```
