 ASCII is a character encoading standard that assigns numerical values to letters, digits , charts 

 In python ord() function converts a single character to its ASCII Value

```python
print(ord('A'))
print(ord('a'))
print(ord('1'))
print(ord(' ')) # space 
print(chr(65))
```

```python
# handling strings:
name = 'Anshuman Maurya'
print([char for char in name]) # list creation
ascii_values = [ord(char) for char in name]
print(ascii_values) 
```
