```python
print('datascience'.isalpha())
print('123'.isalpha())
```

```python
def check_alpha(text):
    c=0
    for i in text:
        if((ord(i) >= 65 and ord(i) <= 90) or (ord(i) >=97 and ord(i) <= 122)):
            c+=1
    if(len(text) == c):
         return True
    else:
        return False
```

```python
print(check_alpha('datascience3'))
print(check_alpha('asddg'))
```

```python
print('1234'.isdigit())
print('machinelearning'.isdigit())
```

```python
def check_digit(text):
    c=0
    for i in text:
        if((ord(i) >= 48 and ord(i) <= 57)): # this will check digit from 0-8=9
            c+=1
    if(len(text)==c):
        return True
    else:
        return False 
```

```python
print(check_digit('datasciene'))
print(check_digit('12344'))
```

```python
def check_digit(text):
    c=0
    for i in text:
        if((ord(i) >= 65 and ord(i) <= 90) or (ord(i) >=97 and ord(i) <= 122) or (ord(i) >= 48 and ord(i) <= 57)):
            c+=1
    if(len(text) == c):
        return True 
    else:
        return False 
```

```python
print(check_digit('as4'))
print(check_digit('asdd'))
```
