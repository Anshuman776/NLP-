# 1.string to ascii conversion

```python
text = "Hey My Name is Anshuman Maurya , I am a third year btech student."
```

```python
print([char for char in text])
ascii_ = [ord(char) for char in text]
print(ascii_)
```

```python
print([{char:ord(char)} for char in text])
```

# 2. Converting ASCII to String

```python
print([chr(asc) for asc in ascii_])
```

```python
''.join([chr(asc) for asc in ascii_])
```
