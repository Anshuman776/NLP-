```python
import re
```

1. Removing the speical characters 

```python
txt = 'Anshuman is happ%y! But, wh$y is h*e not playing?'
```

```python
''.join(re.findall('[^!*%,?$]+',txt))
```

# ^ is used for exclude 

2. Exclusion 

```python
txt = "I'm Anshuman And 20 years old"
print(''.join(re.findall('\d',txt)))
# \D is used for exclusion 
# \d is used for numeric searching 
print(''.join(re.findall('\D',txt)))
```

3. finding patterns link like work-it-out

```python
txt = "Hello I am studing from AKGEC-Ghaziabad-Delhi and it is a amazing . let's see how this work-it-out"
```

```python
# first method brute method 
re.findall('\w\w\w\w-\w\w-\w\w\w',txt)
# in this only we check pattern matvhing exact word by word 
```

```python
 
re.findall('[\w]',txt)
```

```python
# using + sigh combins the word 
re.findall('[\w]+',txt)
```

```python
# second method
re.findall('[\w]+-[\w]+-[\w]+',txt)
# we donot check words length only sees the pattern 
```

4. Phone Number Pattern Matching 

```python
for no in['64-54637-566','543-5432-654','67454-44-763','435-234-6324']:
    print(str(re.findall('[\d]+-[\d]+-[\d]',no)[0]).replace('-',''))
```
