```python
# We will be searching the word present in the text and print the index of the word present within the text. 
# Let us see with the help of the code.
```

```python
txt = "Let us study Data Science from GFG"
```

```python
txt = txt.lower() # convert the string to lowercase
```

```python
word ='data' # Word to search for in the string 
for i in range(len(txt)): #Loop through the string
    if(txt[i:i+len(word)]==word):  # check if the substring matches the word
        print(i,i+len(word)) # Print the start and end indices of the match
     
```

```python
# So we got the index of data present in the text which we desired.

# First, we converted the text to lowercase so that we don’t face case sensitive issues.
# Then we used i:i+len(word) because the extraction would be based on the length of the words.
# Since data is of length 4 the iteration keeps on going till, we get the word.
```
