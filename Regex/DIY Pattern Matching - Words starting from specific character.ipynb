1. Finding Words Starting with Specific Characters

```python
txt = "Ai is having the capability to revolutionize all the industies . The Industry is ready for the change " 
word ='in'
for wrd in txt.split(' '):
    if(wrd[:2].lower() == word.lower()):
        print(wrd)
```

```python
# We define the input string txt.
# We search for words in the string that start with 'in'.
# We use the split() method to break the text into individual words.
# The comparison wrd[:2].lower() == word.lower() ensures that we compare the first two characters of each word in a case-insensitive manner
```

2. Finding Words Ending with Specific Characters

```python
txt = 'AI is having the capability to revolutionize all the industries. The Industry is ready for a big change'
word = 'he'
for wrd in txt.split(' '):
    if(wrd[-2:].lower() == word.lower()):
        print(wrd)
```

```python
# We search for words that end with 'he'.
# The comparison wrd[-2:].lower() == word.lower() ensures that we are checking the last two characters of each word, in a case-insensitive manner.
```

3. Extracting Word Pairs

```python
txt = 'AI is having the capability to revolutionize all the industries. The Industry is ready for a big change'

for i in range(len(txt.split(' '))-1):
    print(txt.split(' ')[i], txt.split(' ')[i+1])
```

```python
# We loop through each word in the text and print the current word and the next word.
# The txt.split(' ')[i], txt.split(' ')[i + 1] expression gives us the current and the next word in the sentence.
```
