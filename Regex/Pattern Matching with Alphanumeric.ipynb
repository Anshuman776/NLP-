```python
import re
print(re.findall('at','The rat sat on the mat and attached by a cat'))
```

```python
print(re.findall('.at','The rat sat on the mat and attached by a cat'))
```

This dot symbol which is being indicated by the arrow here is being used to print one character before the pattern ‘at’.

```python
print(re.findall('at.','The rat sat on the mat and attached by a cat'))
```

```python
print(re.findall('\d','4 is divisible by 2 and not by 3'))
```

Here we have just normally displayed all the numbers present in the text by finding the patterns using \d. Now let us check how to verify the starting or ending character.


```python
print(re.findall('^\d','4 is divisible by 2 and not by 3'))
print(re.findall('^\d','Hi, 4 is divisible by 2 and not by 3'))
```

We can see when we did not have anything before 4 in the text we got our output 4 as it contains a digit starting of the text. 

When we added ‘Hi’ before 4 we can see that we got an empty list because it is not what we desired. As per our condition we needed the text to start with a digit but that is not happening so we our getting an empty list

```python
print(re.findall(r'\d$','4 is divisible by 2 and not by 3'))
print(re.findall(r'\d$','4 is divisible by 2 and not by 3 in maths'))
```

Here we can see as well when we were ending our text with a digit we were getting the output as 3 but when we added ‘in maths’ the text wasn’t ending with a digit due to which we got an empty list as output.

```python
print(re.findall('[A-Z]','Welcome to ANSHUMAN 1'))
print(re.findall('[a-z]','Welcome to ANSHUMAN 1')) 
print(re.findall('[0-9]','Welcome to ANSHUMAN 1')) 
print(re.findall('[A-Za-z]','Welcome to ANSHUMAN 1')) 
print(re.findall('[A-Za-z0-9]','Welcome to ANSHUMAN 1')) 
```
