```python
import nltk
nltk.download('all')
```

```python
txt = "Hello Anshuman . We're hoping you guys are doing great."
print(len(txt.split('.')))  #this gives number of sentences 
print(len(txt.split()))    # this gives number of words
```

# But we can see the number of sentences is not 3, it’s 2 so we can use the nltk library and use the word tokenizer or sentence tokenizer to get the actual number of words and sentences.

```python
from nltk.tokenize import word_tokenize , sent_tokenize
len(sent_tokenize(txt))
```

# so we have got the correct answer 2 because we can see from our eyes its 100% corrrect 



```python
# now if we want to see all the word present in the txt using word_tokenize:
for word in word_tokenize(txt):
    print(word)
```

# now we not want the fullstops(.) being present as word in our word tokenizer so that we can edit the code based on our convenient:

```python
for word in word_tokenize(txt):
    if(word!='.'):
        print(word)
```
