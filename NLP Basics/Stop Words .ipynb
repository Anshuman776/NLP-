```python
import nltk 
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize

txt = 'This is not a good time to talk.'
txt = word_tokenize(txt)
for word in txt:
    print(word)
```

# we have just printed all the words presents in the text using word tokenizer
Now we will print all the words after removing all the stop words present in the 'txt' varible 

```python
# Get the set of english stop words
stop_words = set(stopwords.words('english'))

# Remove stop words from the tokens 
filtered_tokens =[w for w in txt if w.lower() not  in stop_words]
# print the filtered tokens 
for token in filtered_tokens:
    print(token)
```

 this words such as this , is , to , etc  have been removed since they belonged to the category of stop words . we have the condition len(word)>=2 to remove the full stop this condition can be used to remove any kind of punctuation marks
