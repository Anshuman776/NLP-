```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Both Stemming and Lemmatization are used to find the root words but in Lemmatization the root word makes sense and in Stemming the root word might not make sense all the time.
```

```python
from nltk.stem import WordNetLemmatizer , PorterStemmer
stem = PorterStemmer()
lem = WordNetLemmatizer()
```

# Lemmatization


```python
print(lem.lemmatize('change'))
print(lem.lemmatize('changes'))
print(lem.lemmatize('changed'))
```

 # Stemming 

```python
print(stem.stem('change'))
print(stem.stem('changes'))
print(stem.stem('changed'))
```

# The root  word turns out to be the same for stemming although chang does not mean anything so the information loss is huge whereas for Lemmatization the information change is less but the answer for all the words are coming out different 

# It’s  up to us  whether we want to choose Lemmatization or Stemming but in most of the cases Lemmatization is used because Lemmatization will give us a word which makes sense.
